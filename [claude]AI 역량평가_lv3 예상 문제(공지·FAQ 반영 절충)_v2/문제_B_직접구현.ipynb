아래 셀은 dev 40개에 `응답 → extract_json → normalize_record → type_errors` 파이프라인을 돌려 **필드 정확도**를 계산하고, test 30개를 제출 파일로 저장합니다.

````python
# [실행만 하세요]
_extract = extract_json if attempt("5-3 재확인", lambda: extract_json('{"a": 1}')) == {"a": 1} else (lambda raw: None)

def _field_match(a, b, k):
    if k == "medications":
        key = lambda m: (m["name"], None if m["dose_mg"] is None else float(m["dose_mg"]), m["times_per_day"])
        return sorted(map(key, a[k]), key=str) == sorted(map(key, b[k]), key=str)
    return a[k] == b[k]

def _run_5_5():
    assert type_errors({**normalize_record(None), "age": True}) != [], "bool을 정수로 통과시킴"
    assert type_errors({**normalize_record(None), "chief_complaint": "cough"}) == []
    rows, n_err, n_parse_fail = [], 0, 0
    for nid in dev.id:
        obj = _extract(ps.mock_llm(nid))
        n_parse_fail += obj is None
        rec = normalize_record(obj)
        if obj is not None and type_errors(rec):    # 파싱 실패 레코드는 chief_complaint가 null이라 제외
            n_err += 1
        rows.append({k: _field_match(rec, dev_gold[nid], k) for k in SCHEMA_KEYS})
    acc = pd.DataFrame(rows).mean()
    print("필드별 정확도\n", acc.round(3).to_string())
    print(f"전체 필드 정확도 {acc.mean():.3f} | 파싱 실패 {n_parse_fail}건 | 자료형 위반 레코드 {n_err}건")
    print("기준(필드 정확도 ≥ 0.95, 위반 0건):", bool(acc.mean() >= 0.95 and n_err == 0))
    raw_acc = np.mean([[(_extract(ps.mock_llm(n)) or {}).get(k) == dev_gold[n][k] for k in SCHEMA_KEYS] for n in dev.id])
    print(f"(참고) 정규화 없이 응답 그대로 쓸 때 필드 정확도 {raw_acc:.3f}")
    out = [{"id": nid, "record": normalize_record(_extract(ps.mock_llm(nid)))} for nid in test_notes.id]
    ps.save_jsonl(out, "q5_submission.jsonl")
    print("q5 형식 점검:", ps.check_submission("q5") or "통과")
_ = attempt("5-5", _run_5_5)
````

````python
# [실행만 하세요]
import subprocess

def local_score():
    cmd = [sys.executable, "-I", "-X", "utf8", str(ROOT / "정답_채점용" / "score_local.py"), str(ps.SUBMIT), "--json"]
    proc = subprocess.run(cmd, capture_output=True, text=True, encoding="utf-8")
    lines = proc.stdout.strip().splitlines()
    print("\n".join(l for l in lines if not l.startswith("SCORES_JSON")))
    if proc.returncode != 0:
        raise RuntimeError(proc.stderr.strip()[-500:])
    return json.loads(next(l for l in lines if l.startswith("SCORES_JSON"))[len("SCORES_JSON "):])

def compare_with_best(scored, history_path=ps.SUBMIT / "_score_history.json"):
    hist = json.loads(history_path.read_text(encoding="utf-8")) if history_path.exists() else {"best": {}, "runs": []}
    now = time.strftime("%Y-%m-%d %H:%M:%S")
    rows = []
    for kind, r in scored["results"].items():
        new = r["score"] if r["score"] is not None else 0.0          # 형식 오류 → 0점
        old = hist["best"].get(kind, {}).get("score")
        if old is None:
            status = "첫 기록"
        elif new > old + 1e-9:
            status = "▲ best 갱신"
        elif new < old - 1e-9:
            status = "▼ 하락 (best 유지)"
        else:
            status = "= 같음"
        rows.append({"문항": kind, "기준": scored["thresholds"][kind], "기존 best": old, "이번 점수": round(new, 4),
                     "변화": None if old is None else round(new - old, 4), "PASS": r["pass"], "상태": status})
        if old is None or new > old:
            hist["best"][kind] = {"score": new, "pass": r["pass"], "time": now}
    hist["runs"].append({"time": now, "scores": {k: (v["score"] or 0.0) for k, v in scored["results"].items()}})
    history_path.write_text(json.dumps(hist, ensure_ascii=False, indent=1), encoding="utf-8")
    print("\n" + pd.DataFrame(rows).to_string(index=False))
    res = scored["results"]
    if "q4" in res and "q4_ens" in res:
        a, b = res["q4"]["score"] or 0.0, res["q4_ens"]["score"] or 0.0
        print(f"\nQ4 이번 실행: 단일 모델 {a:.4f} → 앙상블 {b:.4f} ({b - a:+.4f})")
    q4_best = max([hist["best"][k]["score"] for k in ("q4", "q4_ens") if k in hist["best"]], default=None)
    if q4_best is not None:
        print(f"Q4 지금까지의 best (단일·앙상블 중): {q4_best:.4f}")
    print(f"채점 {len(hist['runs'])}회째 | 기록: {history_path.relative_to(ROOT)}")

scored = attempt("로컬 채점", local_score)
if scored:
    compare_with_best(scored)
````
